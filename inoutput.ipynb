```python
#1~100 사이의 숫자 중에서 3의 배수만을 더한 후 결과값 출력
hap = 0
num = 1

while num <= 100:
    if num % 3 == 0:
        hap += num
    num += 1

print("1~100 사이의 3의 배수의 합:", hap)
```

```python
#1~100 사이의 숫자 중에서 3의 배수(단 9의 배수 빼기)만을 더한 후 결과값 출력

hap = 0
num = 1

while num <= 100:
    if num % 3 == 0 and num % 9 != 0:
        hap += num
    
    num += 1
print("1~100 사이의 3의 배수(단 9의 배수 빼기)의 합:", hap)
```

```python
#1~100 사이의 숫자 중에서 3의 배수와 5의 배수 더한 후 결과값 출력
hap = 0
num = 1

while num <= 100:
    if num % 3 == 0 or num % 5 == 0:
        hap += num
    
    num += 1
print("1~100 사이의 3의 배수와 5의 배수의 합:", hap)
```

```python
# 주차장 프로그램
# 사용자에게 메뉴 제공
# [1]주차 | [2]출차 | [3]종료

# 사용자는 메뉴 선택
# 메뉴 1번 클릭 : 1 입력
# 자동차 A ~ E, 자동차 번호는 들어오는 순서대로 A부터 E까지 부여됨
# A 자동차 입차, 주차장 상태 ==> ['A']
# B 자동차 입차, 주차장 상태 ==> ['A','B']
# 메뉴 2번 클릭
# 출입구가 하나 밖에 없는 스택 형태
# B 자동차 출차, 주차장 상태 ==> ['A']


# 주차장 상태 ==> ['A','B','C','D','E'] => 꽉참
# 메뉴 1번 클릭 :  1 입력
# 자동차 A ~ E
# 주차장에 자리가 없음 출력




parking_lot = []
top, car_name = 0, 'A'

while True:
    no = int(input("[1] 주차 | [2] 출차 | [3] 종료"))
    if no <= 3:
        print()
    else:
        parking_lot.append(car_name)
        print(f"{car_name} 자동차입차. 주차장 상태 ==> {parking_lot}")
        car_name = chr(ord(car_name)+1)
        top = top + 1
    elif no == 2:
        if top > 0;
            outCar = parking_lot.pop() 
            print(f"{outCar} 자동차 출차. 주차장 상태 ==>{parking_lot}")
            top = top -1
            

    
```

```python
# 주차장 프로그램
# 사용자에게 메뉴 제공
# [1]주차 | [2]출차 | [3]종료

# 사용자는 메뉴 선택
# 메뉴 1번 클릭 : 1 입력
# 자동차 A ~ E, 자동차 번호는 들어오는 순서대로 A부터 E까지 부여됨
# A 자동차 입차, 주차장 상태 ==> ['A']
# B 자동차 입차, 주차장 상태 ==> ['A','B']
# 메뉴 2번 클릭
# 출입구가 하나 밖에 없는 스택 형태
# B 자동차 출차, 주차장 상태 ==> ['A']


# 주차장 상태 ==> ['A','B','C','D','E'] => 꽉참
# 메뉴 1번 클릭 :  1 입력
# 자동차 A ~ E
# 주차장에 자리가 없음 출력



parking = []      
car = "A"         

while True:
    print("\n[1] 주차  [2] 출차  [3] 종료")
    menu = input("메뉴 선택 : ")

    if menu == "1":
        if len(parking) >= 5:
            print("주차장에 자리 없음")
        else:
            parking.append(car)
            print("주차장 상태 :", parking)
            car = chr(ord(car) + 1) 

    elif menu == "2": 
        if len(parking) == 0:
            print("출차할 자동차가 없음")
        else:
            out_car = parking.pop()
            print("주차장 상태 :", parking)

    elif menu == "3":   
        print("종료")
        break

    else:
        print("잘못된 입력")
```

```python
print([i for i in range(1,7)])
print([i * i for i in range(5)])
print([i * 2 for i in [1,3,7,9,16]])
print([[i, i + 1, i+2 ]for i in [1,3,7,9,16]])

# 1~ 10 사이의 숫자 중에서 홀수 리스트 생성
odd_list = []
for i  in range(1,11):
    if i % 2 == 1:
        odd_list.append(i)
odd_list

odd_list2 = [i for i  in range(1,11) if i % 2 == 1]

```

```python
list1 = ["갑","을","병","정","경"]
list2 = []
# q_list 에서 정을 제외한 요소를 갖는 새로운 리스트 생성
#for
for i in list1:
    if i != "정":
        list2.append(i)
print(list2)


# 리스트 컴프리헨션
print([i for i in ["갑","을","병","정","경"] if i !="정"])
print([i for i in list1 if i !="정"])
```

```python
q1 = ['nice', 'study', 'python', 'anaconda', '!']
# 5글자 이상의 단어만 새로운 리스트로 생성
#  len()
q1_1 = []
for i in q1_1:
    if len(i) >=5:
        q1_1.append(i)
print(q1_1)

```

```python
q2 = ['A','b','c','D','e','F','G','h']
# 소문자만 추출 후 새로운 리스트로 생성
# "x".islower()  => x 가 소문자가 맞는가
# "x".isupper()  => x 가 대문자가 맞는가
q2_2 = []
for i in q2:
    if i.islower():
        q2_2.append(i)
print(q2_2)
# 리스트 컴프레헨션으로 표현
[i for i in q2 if i.islower()]
```
