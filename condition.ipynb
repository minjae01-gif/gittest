```python
# https://naver.com
#비밀번호 생성
#규칙 1. https:// 제외
#규칙 2. 처음 만나는 . 이후 부분은 제외
#규칙 3. 남은 글자 중 처음 세자리 + 글자 갯수 + 글자 내 'e' 개수 + '!' 비밀번호 생성

# 문자함수, 슬라이싱, str()


url = "https://naver.com"
str_n = url.replace("https://", "")
str_n = str_n[:str_n.find(".")]
#str_n = str_n[:str_n.index(".")] //index는 못찾으면 에러남
psw = str_n[:3] + str(len(str_n)) + str(str_n.count("e")) + "!"

print(psw)

```

```python
#2026년이 윤년 or  평년 인지 판별

year = 2026
# 윤년 조건
# 년도를 4로 나눈 나머지가 0 이고, 년도를 100으로 나눈 나머지가
# 0이 아니거나 년도를 400으로 나눈 나머지가 0 이면

if year % 4 == 0 and ((year % 100 != 0) or (year % 400 == 0)):
    print("윤년")
else:
    print("윤년 아님")


```

```python
#남자(1,3), 여자(2,4) 판별
jumin = "001205-3234567"

gender = jumin.split("-")[1][0]

if gender in ("1", "3"):
    print("남자")
elif gender in ("2", "4"):
    print("여자")
else:
    print("오류")
```

```python
# 한 시간 임금은 10700원, 8시간 까지는 10700 으로 지금하고 8시간 초과 근무시에는 1.5배 지급
# hours(근무시간), rate(시간당임금), pay(일당)

hours, rate, pay = 12, 10700, 0

if(hours > 8):
    pay = (8 * rate) + (hours-8) * rate * 1.5
elif(hours <=8):
    pay = hours * rate
else:
    print("오류")
print(pay)
print("근무시간: %d, 일당: %d" % (hours, pay))
print("근무시간: {0}, 일당: {1:.0f}".format(hours, pay))
print(f"근무시간: {hours}, 일당: {pay:.0f}")

```
